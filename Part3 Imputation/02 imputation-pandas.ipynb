```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'age': [25, 32, None, 40, 28],
    'income': [50000, 60000, 55000, None, 70000],
    'gender': ['female', None, None, 'male', 'male']
})

df
```

```python
# Median imputation
df['age'].fillna(df['age'].median())
```

```python
# Mean imputation
df['income'].fillna(df['income'].mean())
```

```python
# Arbitrary value for numeric column
df['income'].fillna(-999)
```

```python
# Arbitrary value for categorical column
df['gender'].fillna('female')
```

```python
# Mode / Most Frequent Category Imputation
df['gender'].fillna(df['gender'].mode()[0])
```

```python
# Complete Case Analysis (CCA): Remove rows where any or specific columns have missing values.

# Drop rows with any missing values
df_clean = df.dropna()

# Drop rows with missing values in specific columns
df_clean = df.dropna(subset=['age', 'income'])
```

```python
# End of Distribution Imputation: Replace missing values with extreme values in the tail (min or max ± epsilon).

# Using a high value for income
df['income'].fillna(df['income'].max() + 1000)

# Using a low value
df['age'].fillna(df['age'].min() - 10)
```

```python
# Random Sample Imputation

df['age'].fillna(np.random.choice(df['age'].dropna()))
```
