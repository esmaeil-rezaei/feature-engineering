### Handling Missing Data in Machine Learning

**Mean or Median Imputation**  
Use when data is **missing completely at random** and **no more than 5% of the variable contains missing values**.

**Pros:**

- Simple to implement
- Median imputation is robust to outliers
- Preserves the central tendency of the data

**Cons:**

- Reduces variability in the data
- Can introduce bias if missingness is not random

---

**Arbitrary Value Imputation**  
Use when data is **not missing at random** and you want to flag missing observations rather than replace with a central value.

**Pros:**

- Lets the model detect patterns in missingness
- Widely used in practice

**Cons:**

- Choosing the appropriate value is challenging
- Extreme values may distort relationships in some models

---

**Mode / Most Frequent Category Imputation**  
Use for categorical variables when data is **missing completely at random** and **no more than 5% of variables are missing**.

**Pros:**

- Simple to implement
- Preserves the most common category

**Cons:**

- Can introduce bias if missingness is systematic
- Frequent categories may dominate the distribution

---

**Complete Case Analysis (CCA)**  
Use when data is **missing completely at random** and **no more than 5% of the total dataset is missing**.

**Pros:**

- Simple and avoids imputation bias

**Cons:**

- Discards data, reducing statistical power
- Can bias results if missingness is not completely random

---

**End of Distribution Imputation**  
Typically replaces missing values with extreme values at the tail of the distribution.

**Pros:**

- Flags missing values explicitly for algorithms that can interpret outliers

**Cons:**

- Can distort the distribution if applied incorrectly

---

**Random Sample Imputation**  
Use when data is **missing completely at random** and **no more than 5% of variables are missing**.

**Pros:**

- Preserves the variability of the original distribution
- Works well with linear models
- Can be combined with missing indicators to leverage missingness patterns

**Cons:**

- Introduces randomness, which may increase variance across runs
