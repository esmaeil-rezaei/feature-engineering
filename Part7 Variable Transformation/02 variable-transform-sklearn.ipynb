```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import (
    MinMaxScaler, StandardScaler, Normalizer, FunctionTransformer,
    PowerTransformer, PolynomialFeatures, KBinsDiscretizer, QuantileTransformer
)
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
```

```python
data = {
    'Age': [18, 22, 25, 35, 45, 50, 60, 70],
    'Income': [15000, 20000, 25000, 50000, 80000, 90000, 120000, 150000],
    'Debt': [1000, 1500, 2000, 5000, 8000, 9000, 12000, 15000],
    'Score': [50, 55, 60, 70, 80, 85, 90, 95]
}
df = pd.DataFrame(data)

df
```

### 1. Scaling


```python
scaler = MinMaxScaler()
scaled = scaler.fit_transform(df[['Age', 'Income', 'Debt']])
pd.DataFrame(scaled, columns=['Age', 'Income', 'Debt'])
```

```python
std_scaler = StandardScaler()
standardized = std_scaler.fit_transform(df[['Age', 'Income', 'Debt']])
pd.DataFrame(standardized, columns=['Age', 'Income', 'Debt'])
```

### 2. Normalization


```python
normalizer = Normalizer()
normalized = normalizer.fit_transform(df[['Age', 'Income', 'Debt']])
pd.DataFrame(normalized, columns=['Age', 'Income', 'Debt'])
```

### 3. Log / Power Transform


```python
log_transformer = FunctionTransformer(np.log1p)  # log(1 + x)
log_transformed = log_transformer.fit_transform(df[['Income', 'Debt']])
pd.DataFrame(log_transformed, columns=['Income', 'Debt'])
```

```python
sqrt_transformer = FunctionTransformer(np.sqrt)
sqrt_transformed = sqrt_transformer.fit_transform(df[['Income', 'Debt']])
pd.DataFrame(sqrt_transformed, columns=['Income', 'Debt'])
```

### 4. Box-Cox Transform


```python
# Box-Cox requires strictly positive values
boxcox = PowerTransformer(method='box-cox')
boxcox_transformed = boxcox.fit_transform(df[['Income', 'Debt']])
pd.DataFrame(boxcox_transformed, columns=['Income', 'Debt'])
```

### 5. Yeo-Johnson Transform


```python
# Yeo-Johnson works with zero or negative values
yeojohnson = PowerTransformer(method='yeo-johnson')
yeojohnson_transformed = yeojohnson.fit_transform(df[['Income', 'Debt']])
pd.DataFrame(yeojohnson_transformed, columns=['Income', 'Debt'])
```

### 6. Polynomial Features


```python
poly = PolynomialFeatures(degree=2, include_bias=False)
poly_features = poly.fit_transform(df[['Age', 'Income']])
poly_columns = poly.get_feature_names_out(['Age', 'Income'])
pd.DataFrame(poly_features, columns=poly_columns)
```

### 7. Binning / Discretization


```python
kbins = KBinsDiscretizer(n_bins=3, encode='ordinal', strategy='uniform')
binned = kbins.fit_transform(df[['Score']])
pd.DataFrame(binned, columns=['Score_Bin'])
```

### 8. Quantile / Rank Transform


```python
quantile = QuantileTransformer(output_distribution='normal')
quantile_transformed = quantile.fit_transform(df[['Income', 'Debt']])
pd.DataFrame(quantile_transformed, columns=['Income', 'Debt'])
```
