# Variable Transformation in Machine Learning

In machine learning, **variable transformation** (or **feature transformation**) modifies variables to improve model performance, interpretability, or stability. Raw data often contains variables that are **skewed, non-linear, or have different scales**, which can negatively affect many machine learning algorithms. Transforming variables can help in several ways:

### Why Variable Transformation is Needed

1. **Normalize Scales**: Features may have very different ranges. Algorithms like gradient descent, k-nearest neighbors, or regularized regression perform better when features are on similar scales.
2. **Reduce Skewness**: Highly skewed variables can dominate the learning process or lead to biased predictions.
3. **Linearize Relationships**: Many models assume linear relationships between features and target. Transforming variables can help satisfy this assumption.
4. **Improve Model Stability**: Transformations can reduce heteroscedasticity (non-constant variance) and improve statistical properties.
5. **Handle Non-Normality**: Some algorithms (e.g., linear regression) assume residuals are normally distributed. Transforming variables can help approximate normality.

# Common Variable Transformers

Variable transformation methods can be broadly classified into:

1. **Scaling and Standardization**
2. **Normalization**
3. **Logarithmic and Power Transforms**
4. **Box-Cox and Yeo-Johnson Transforms**
5. **Polynomial Features**
6. **Binning / Discretization**
7. **Rank / Quantile Transform**

---

## 1. Scaling and Standardization

- **Definition:** Rescale numerical variables to have a specific scale.
- **Techniques:**
  - **Min-Max Scaling**: Rescales features to a fixed range, usually [0, 1].
    $$
    X_{scaled} = \frac{X - X_{min}}{X_{max} - X_{min}}
    $$
  - **Standardization (Z-score)**: Centers features to mean 0 and variance 1.
    $$
    X_{std} = \frac{X - \mu}{\sigma}
    $$
- **Use Case:** Gradient-based models, k-NN, SVM, PCA.

---

## 2. Normalization

- **Definition:** Scales feature vectors to have **unit norm** (length = 1).
- **Techniques:** L1, L2, or max normalization.
- **Use Case:** Distance-based algorithms like k-NN or clustering.

---

## 3. Logarithmic and Power Transforms

- **Definition:** Apply functions like $\text{log}(x)$, $\text{sqrt}(x)$, or $x^2$ to reduce skewness or linearize relationships.
- **Example:**
  - Log-transform positive skewed features: $X_{new} = \text{log}(X + 1)$
  - Square-root transform: $X_{new} = \text{sqrt}(X)$
- **Use Case:** Highly skewed variables such as income, population, or counts.

---

## 4. Box-Cox and Yeo-Johnson Transforms

- **Box-Cox Transform:** Converts **positive numeric data** to approximate normality.

  $$
  y(\lambda) =
    \begin{cases}
      \frac{y^\lambda - 1}{\lambda} & \lambda \neq 0 \\
      \log(y) & \lambda = 0
    \end{cases}
  $$

  - **Limit:** Can only be applied to strictly positive values.

- **Yeo-Johnson Transform:** Generalization of Box-Cox that works for **zero and negative values**.
  $$
  y^{(\lambda)} =
    \begin{cases}
      ((y + 1)^\lambda - 1)/\lambda & y \ge 0, \lambda \neq 0 \\
      \log(y + 1) & y \ge 0, \lambda = 0 \\
      -((-y + 1)^{2-\lambda} - 1)/(2-\lambda) & y < 0, \lambda \neq 2 \\
      -\log(-y + 1) & y < 0, \lambda = 2
    \end{cases}
  $$
- **Use Case:** When features contain **negative or zero values** and you want approximate normality for linear models.

---

## 5. Polynomial Features

- **Definition:** Create new features by raising existing features to a power or multiplying features together.
- **Example:** If $X_1$ and $X_2$ are features, polynomial features include: $X_1^2$, $X_2^2$, $X_1 \times X_2$.
- **Use Case:** Capture non-linear relationships in linear models.

---

## 6. Binning / Discretization

- **Definition:** Convert continuous variables into categorical bins.
- **Techniques:** Equal-width, equal-frequency, or custom thresholds.
- **Use Case:** Tree-based models, handling skewed features, or creating categorical features for interpretability.

---

## 7. Rank / Quantile Transform

- **Definition:** Transform features based on their **rank or cumulative distribution**.
- **Example:** Quantile transformer maps data to a uniform or normal distribution.
- **Use Case:** Reduce the effect of outliers, enforce normality.

---

### Summary

| Transformer Type          | Purpose                                      | Use Case                                  |
| ------------------------- | -------------------------------------------- | ----------------------------------------- |
| Scaling / Standardization | Rescale to common range / mean 0 variance 1  | Gradient-based models, PCA                |
| Normalization             | Unit vector length                           | k-NN, clustering                          |
| Log / Power               | Reduce skewness, linearize                   | Skewed numeric features                   |
| Box-Cox                   | Positive numeric, approximate normality      | Linear models, variance stabilization     |
| Yeo-Johnson               | Handles zero/negative, approximate normality | Linear models with negative/zero features |
| Polynomial                | Capture non-linear relationships             | Linear regression                         |
| Binning                   | Convert numeric to categorical               | Tree models, interpretability             |
| Rank / Quantile           | Reduce outlier effect, enforce normality     | Robust modeling, preprocessing            |
