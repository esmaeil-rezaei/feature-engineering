## Types of Variables

In feature engineering, variable types determine what you are allowed to do with the data. They control which transformations are valid, how models interpret features, and how information is encoded.

Below are the main categories of variables commonly encountered in feature engineering:

**1. Numerical**

- _Definition:_ Variables that represent measurable quantities and can be used in arithmetic operations.
- _Example:_ Age (30), Salary (50000), Temperature (22.5°C)

**2. Categorical**

- _Definition:_ Variables that represent discrete groups or categories, usually without any inherent order (nominal) or with an order (ordinal).
- _Example:_ Gender (Male, Female), Education Level (High School, Bachelor, Master)

**3. Date/Time**

- _Definition:_ Variables that represent dates or timestamps, which can be used to extract temporal features like day, month, or season.
- _Example:_ Order Date (2025-12-01), Timestamp (2025-12-01 14:30:00)

**4. Boolean**

- _Definition:_ Variables that have only two possible values: True or False (1 or 0).
- _Example:_ Is_Active (True/False), Has_Purchased (1/0)

**5. Text**

- _Definition:_ Variables that contain unstructured textual data, which often require special processing like tokenization or embedding.
- _Example:_ Product Review ("Great product!"), Tweet ("Happy New Year!")

**6. Mix**

- _Definition:_ Variables that combine multiple types of data or require multiple transformations, such as lists or complex structures.
- _Example:_ Address ("123 Main St, Apt 4, Boston, MA"), Sensor Data ([23.5, 24.1, 22.8])


```python
from datetime import datetime
import pandas as pd

def detect_feature_type(value):
    """
    Detect the type of a variable for feature engineering purposes.
    """
    if isinstance(value, bool):
        return "Boolean"
    elif isinstance(value, (int, float)):
        return "Numerical"
    elif isinstance(value, str):
        # Try to see if it can be parsed as a date
        try:
            pd.to_datetime(value)
            return "Date/Time"
        except:
            return "Text"  # If not date, it's text
    elif isinstance(value, (list, tuple, dict)):
        return "Mix"
    elif isinstance(value, datetime):
        return "Date/Time"
    else:
        return "Unknown"
```

```python
examples = [
    30,                            # Numerical
    50000.0,                       # Numerical
    "Male",                        # Categorical/Text
    "2025-12-01",                  # Date/Time
    True,                          # Boolean
    "Great product!",              # Text
    ["123 Main St", 4, "Boston"],  # Mix
]
```

```python
for val in examples:
    feature_type = detect_feature_type(val)
    print(f"Value: {val} -> Feature Type: {feature_type}")
```
