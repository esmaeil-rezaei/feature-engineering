## General example

Here is a table of all variable types, which we will use as a basis to explain the different types of variables and how to work with them.


```python
import pandas as pd

data = {
    "Age": [25, 30, 22, 28, 35, 40, 19, 31, 27, 33],  # Numerical
    "Gender": ["Male", "Female", "Female", "Male", "Male", 
               "Female", "Male", "Female", "Male", "Female"],  # Categorical
    "Num_Children": [0, 2, 3, 1, 0, 0, 2, 2, 3, 1],  # Categiorical
    "Signup_Date": [
        "2025-01-01", "2025-01-03", "2025-01-05", "2025-01-07", 
        "2025-01-09", "2025-01-11", "2025-01-13", "2025-01-15",
        "2025-01-17", "2025-01-19"
    ],  # Date/Time
    "Is_Active": [True, False, True, False, True, False, True, True, False, True],  # Boolean
    "Review": [
        "Great product!", "Needs improvement", "Loved it", "Not bad", 
        "Excellent", "Could be better", "Highly recommend", "Average", 
        "Terrible experience", "Will buy again"
    ],  # Text
    "Address": [
        "123 Main St, Apt 4, Boston, MA", "456 Oak St, Apt 12, NY", 
        "789 Pine St, Apt 3, Chicago, IL", "321 Maple St, Apt 9, SF, CA", 
        "654 Cedar St, Apt 1, Seattle, WA", "987 Elm St, Apt 5, Austin, TX", 
        "147 Birch St, Apt 2, Miami, FL", "258 Walnut St, Apt 8, Denver, CO", 
        "369 Chestnut St, Apt 7, Portland, OR", "159 Spruce St, Apt 6, Atlanta, GA"
    ]  # Mix
}

df = pd.DataFrame(data)
df
```

We can quickly check the data types of the variables in a DataFrame using `dtypes`.


```python
df.dtypes
```

## Numrical

### Type Mismatches

Feature stored as one type (int) but used as another (float). For example, `Age` stored as integer, but during feature engineering you want to normalize it as a decimal. So, you need to convert the type to float in this use case.


```python
df["Age"] = df["Age"].astype(float)
print(df["Age"].dtype)
```

To revert a variable to an integer type, you can use


```python
df["Age"] = df["Age"].astype(int)
print(df["Age"].dtype)
```

### Implicit Categorical Numbers

Some numbers are categorical in disguise. `Number of children = 0,1,2,3` — small integers, but not continuous. Treating them as continuous can distort scaling or polynomial features.


```python
df["Num_Children"] = df["Num_Children"].astype("category")
print(df["Num_Children"].dtype)
```

## Categorical

For analysis, we usually convert categorical variables (text-based) into numbers. There are two common methods:

1. **Ordinal Encoding:** Assigns an integer to each category, preserving a meaningful order (if one exists). We use it when the categories have a natural ranking.


```python
from sklearn.preprocessing import OrdinalEncoder

data = {"Education": ["High School", "Bachelor", "Master", "PhD"]}
df_cat = pd.DataFrame(data)

ordinal_encoder = OrdinalEncoder()
df_cat["Education_Encoded"] = ordinal_encoder.fit_transform(df_cat[["Education"]])
df_cat
```

**2. One-Hot Encoding:** Converts each category into a separate binary column (0/1 or True/False), with no ordinal relationship implied. It is used when categories are nominal (no natural order).


```python
df_cat = pd.DataFrame({"Gender": ["Male", "Female", "Female", "Male"]})

df_onehot = pd.get_dummies(df_cat, columns=["Gender"])
df_onehot
```

### Map manually

Assign integers to categories manually using a dictionary when categories are few, and you know the mapping ahead of time. For example, Converting Gender to numbers (Male=1, Female=0)


```python
df["Gender_encoded"] = df["Gender"].map({'Male': 0, 'Female': 1})
df[["Gender", "Gender_encoded"]]
```

## Datetime

**1. Basic Conversion:** Convert a column of strings representing dates into pandas datetime format. This is prerequisit for manipulate datetime features otherwise python will treat these as string.


```python
# Example column
df['Signup_Date'] = pd.to_datetime(df['Signup_Date'])
df['Signup_Date']
```

**2. Extract Components:** Extract year, month, day, weekday, and hour from a datetime column.


```python
df['Year'] = df['Signup_Date'].dt.year
df['Month'] = df['Signup_Date'].dt.month
df['Day'] = df['Signup_Date'].dt.day
df['Weekday'] = df['Signup_Date'].dt.day_name()
df['Hour'] = df['Signup_Date'].dt.hour

df[["Signup_Date", "Year", "Month", "Day", "Weekday", "Hour"]]
```

**3. Date Differences:** Calculate the difference between two dates in days, months, or years.


```python
df['Days_Since_Signup'] = (pd.Timestamp('2025-12-01') - df['Signup_Date']).dt.days
df[['Signup_Date', 'Days_Since_Signup']]
```

To calculate the difference between two datetime columns, you can subtract them directly. For example: `df['col1'] - df['col2']`.


```python
# Current Age

df_age = pd.DataFrame({
    "Birthday": [
        "1990-05-14",
        "1985-11-02",
        "2000-01-25",
        "1997-07-09",
        "1993-03-18",
        "1988-12-30",
        "2002-09-05",
        "1995-04-21",
        "1980-06-11",
        "1999-10-27"
    ]
})

df_age["Birthday"] = pd.to_datetime(df_age["Birthday"])

df_age['Current_Age'] = (pd.Timestamp.now() - df_age['Birthday']).dt.days
df_age[['Birthday', 'Current_Age']]
```

**4. Filtering by Date:** Select rows within a specific date range.


```python
df_filtered = df[(df['Signup_Date'] >= '2025-01-05') & (df['Signup_Date'] <= '2025-01-15')]
df_filtered
```

**5. Handling Missing Dates:** Fill missing datetime values with a default value or forward/backward fill.


```python
signup_df = pd.DataFrame({
    "Signup_Date": [
        "2025-01-01",
        None,
        "2025-01-05",
        None,
        "2025-01-10",
        "2025-01-12",
        None,
        "2025-01-18",
        None,
        "2025-01-20"
    ]
})

signup_df['Signup_Date'] = pd.to_datetime(signup_df['Signup_Date'])

signup_df['Signup_Date'] = signup_df['Signup_Date'].fillna(pd.Timestamp('2025-01-01'))
signup_df['Signup_Date']
```

```python
# or forward fill
signup_df['Signup_Date'] = signup_df['Signup_Date'].ffill()
signup_df['Signup_Date']
```

**6. Sorting by Date:** Sort the DataFrame by datetime column ascending or descending.


```python
df_sorted = signup_df.sort_values(by='Signup_Date', ascending=True)
df_sorted[['Signup_Date']]
```

**7. Time Delta Features:** Create features like day of week, week of year, quarter, or is_weekend.


```python
df['Week_of_Year'] = df['Signup_Date'].dt.isocalendar().week
df['Quarter'] = df['Signup_Date'].dt.quarter
df['Is_Weekend'] = df['Signup_Date'].dt.weekday >= 5  # Saturday=5, Sunday=6

df[['Signup_Date', 'Week_of_Year', 'Quarter', 'Is_Weekend']]
```

**8. Lag / Rolling Features:** Compute previous date difference, moving averages, or rolling sums.

**`rolling(window=5)`**

- it looks at the _current value + the previous 4 values_
- the window moves down _one row_ at a time

**`.diff()`**

- Computes the difference between the current row and the previous row.
- `diff(periods=2)` finds the difference between the current date and the date two rows before.


```python
df['Days_Since_Last'] = df['Signup_Date'].diff().dt.days
df['7Day_Rolling'] = df['Days_Since_Last'].rolling(window=5).mean()

df[['Signup_Date', 'Days_Since_Last', '7Day_Rolling']]
```

**9. Extract Part of Timestamp:** Extract hour, minute, second from datetime (for event timestamp).


```python
df_dt = pd.DataFrame({
    "Event_Time": [
        "2025-01-01 08:15:30",
        "2025-01-01 09:45:10",
        "2025-01-01 12:00:05",
        "2025-01-02 14:20:50",
        "2025-01-02 16:35:40",
        "2025-01-03 18:10:25",
        "2025-01-03 20:55:00",
        "2025-01-04 22:30:15",
        "2025-01-05 06:05:45",
        "2025-01-05 07:50:35"
    ]
})

df_dt["Event_Time"] = pd.to_datetime(df_dt["Event_Time"])
```

```python
df_dt["Hour"] = df_dt["Event_Time"].dt.hour
df_dt["Minute"] = df_dt["Event_Time"].dt.minute
df_dt["Second"] = df_dt["Event_Time"].dt.second

df_dt
```

**10. Time Zone Conversion:** Convert naive datetime to a specific timezone.


```python
df_dt['Signup_Date_TZ'] = df_dt['Event_Time']\
    .dt.tz_localize('UTC')\
        .dt.tz_convert('US/Eastern')
df_dt
```

**12. Mixed Date and Datetime:** Sometimes, a single variable may contain values in both date and datetime formats. If we try to convert them directly to a single datetime format, it can raise an error. To handle this, we can use `errors='coerce'`, which converts any incompatible values to `NaT` — pandas’ equivalent of NaN for datetime objects.


```python
mixed_df = pd.DataFrame({
    "Event": [
        "2025-01-01",
        None,
        "2025-01-05 08:15:30",
        None,
        "2025-01-10",
        "2025-01-12 20:55:00",
        None,
        "2025-01-18 06:05:45",
        None,
        "2025-01-20"
    ]
})
```

```python
pd.to_datetime(mixed_df["Event"], errors='coerce')
```

The solution is to format all values consistently using `format='mixed'`. Since the column contains null values, we should apply this formatting only to the non-null entries to avoid errors.


```python
mixed_df = pd.to_datetime(mixed_df["Event"], format='mixed')
mixed_df
```
