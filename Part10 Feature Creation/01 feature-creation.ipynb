# Feature Creation Methods

Below is the list of all feature creation methods that we have either **covered in other notebooks** or are **beyond the scope of this tutorial**.  
In this tutorial, we just give an example for **aggregation**.

## Encoding / Transformation

- Label / Ordinal encoding
- One-hot encoding
- Target / Frequency encoding
- Binary encoding

## Date & Time Features

- Year, month, day, weekday
- Hour, minute, second
- Duration / Time difference
- Cyclical encoding (sine/cosine)

## Text / NLP Features

- Word / character counts
- TF-IDF / Bag-of-Words
- Sentiment analysis
- Keyword / Regex flags
- Named Entity Recognition (NER)

## Dimensionality Reduction

- PCA / Kernel PCA
- LDA
- t-SNE / UMAP (for feature visualization)
- Variance thresholding

## Interaction / Polynomial Features

- Pairwise product or ratio
- Polynomial expansion
- Splines / piecewise polynomials

## Model-based Features

- Tree leaf indices
- Gradient boosting interactions
- Feature importances to create new flags

## Aggregation / Grouped Features

- Row-wise: sum, mean, min, max, std
- Group-wise: mean per group, max per group
- Cumulative or expanding statistics

## Mixed / Complex Features

- Extract numbers / flags from text
- Composite features decomposition
- Handling unstructured cells with regex

## Binning / Discretization

- Equal-width / equal-frequency
- Arbitrary / domain-specific bins
- Decision tree / KMeans binning
- Binarization / thresholds

## Custom / Domain-Specific Features

- Ratios, percentages, indices
- Health metrics (BMI, risk scores)
- Finance metrics (ratios, volatility)
- Engineering units conversion


```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "score1": [10, 20, 30, 40],
    "score2": [5, 10, 15, 20],
    "score3": [2, 4, 6, 8]
})

features = ["score1", "score2", "score3"]
```

### Aggregating Multiple Functions in One Line


```python
math_func = ["sum", "prod", "mean", "std", "max", "min"]

df_t = df[features].agg(math_func, axis=1)
df_t.head()
```
