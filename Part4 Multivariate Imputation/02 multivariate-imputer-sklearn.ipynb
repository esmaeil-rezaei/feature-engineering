# Multivariate Imputation Examples in Python

We will show examples of handling missing values using **KNN imputation** and **MICE** with `sklearn`.


```python
import numpy as np
import pandas as pd
# IterativeImputer is considered experimental in some versions (<=0.24)
# and needs a special import before you can use it.
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import KNNImputer, IterativeImputer

data = {
    'Age': [25, 30, np.nan, 35, 40],
    'Salary': [50000, 60000, 65000, np.nan, 80000],
    'Experience': [2, 5, 7, 10, np.nan]
}
df = pd.DataFrame(data)
df
```

### KNN Imputation


```python
knn_imputer = KNNImputer(n_neighbors=2)
df_knn_imputed = knn_imputer.fit_transform(df)

df_knn_imputed = pd.DataFrame(df_knn_imputed, columns=df.columns)
print("\nKNN Imputed Data:")
df_knn_imputed
```

- Good for datasets where similar rows are expected to have similar values.

### MICE Imputation (IterativeImputer)


```python
mice_imputer = IterativeImputer(max_iter=10, random_state=0)
df_mice_imputed = mice_imputer.fit_transform(df)

df_mice_imputed = pd.DataFrame(df_mice_imputed, columns=df.columns)
print("\nMICE Imputed Data:")
df_mice_imputed
```

- Generates more statistically robust imputations, especially when variables are correlated.
