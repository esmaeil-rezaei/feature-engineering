# Multivariate Imputation

In real-world datasets, missing values are common and can bias analyses if not handled properly. **Multivariate imputation** is a technique where missing values in a variable are predicted using the information from **other variables** in the dataset. Unlike simple methods such as mean or median imputation, multivariate imputation considers the relationships between variables, leading to more accurate and realistic imputations.

## K-Nearest Neighbors (KNN) Imputation

**KNN imputation** fills missing values by looking at the *k most similar rows* (neighbors) in the dataset:

- **Idea:** If a value is missing, find the k rows most similar based on other features, and use their values to impute the missing entry.
- **How it works:**
  1. Compute the distance (e.g., Euclidean) between rows with missing values and all other rows.
  2. Identify the k closest rows.
  3. Impute the missing value using the mean (for numeric) or mode (for categorical) of these neighbors.
- **Pros:** Simple, non-parametric, works well with non-linear relationships.
- **Cons:** Computationally expensive for large datasets and sensitive to the choice of k and distance metric.

## MICE (Multiple Imputation by Chained Equations)

**MICE** is a more sophisticated approach that generates multiple imputations for missing values by iteratively modeling each variable as a function of the others:

- **Idea:** Treat each variable with missing values as a regression problem and predict its missing values using other variables.
- **How it works:**
  1. Initialize missing values with a simple method (like mean imputation).
  2. Select one variable with missing data as the target and use other variables as predictors to fit a regression model.
  3. Predict missing values for this variable.
  4. Repeat the process for all variables with missing data (this is one iteration or "cycle").
  5. Repeat cycles multiple times to stabilize imputations.
  6. Optionally, create multiple complete datasets for statistical analysis.
- **Pros:** Accounts for uncertainty by generating multiple plausible values, handles complex relationships between variables.
- **Cons:** More computationally intensive and requires careful model specification.
