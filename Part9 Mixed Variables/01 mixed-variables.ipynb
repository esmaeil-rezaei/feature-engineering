# Mixed Variables

**Mixed variables** are variables in which a single feature contains **more than one type of information**, typically a combination of **numerical values, text, symbols, or units** within the same cell. These variables are common in real-world datasets such as healthcare records, surveys, logs, and administrative data.

Unlike purely numerical or categorical variables, mixed variables **cannot be directly used in statistical analysis or machine learning models**. They must first be transformed into structured representations.

## Structured vs. Unstructured Mixed Variables

### Structured Mixed Variables

**Structured mixed variables** follow a **consistent and predictable format** across observations.  
Although they contain multiple data types, the pattern is regular, making them relatively easy to parse.

**Characteristics:**

- Consistent delimiters (e.g., `/`, `-`)
- Fixed order of components
- Same unit or label across rows

**Example formats:**

- `"120/80"`
- `"50 mg"`
- `"5 years"`

### Unstructured Mixed Variables

**Unstructured mixed variables** do not follow a consistent format.  
The same concept may be expressed using different words, symbols, or orders.

**Characteristics:**

- Inconsistent syntax
- Optional words or symbols
- Missing or implicit information
- Hard to parse using simple string operations

**Example formats:**

- `"~70kg"`, `"about 65 kg"`, `"70-ish"`
- `"less than 2 days"`, `"overnight"`
- `"BP stable, ~130s/80s"`


### Blood Pressure (Structured)


```python
import pandas as pd

df = pd.DataFrame({
    "bp": ["120/80", "140/90", "130/85"]
})

df[["systolic", "diastolic"]] = (
    df["bp"]
    .str.split("/", expand=True)
    .astype(int)
)

df
```

### Structured Mixed Variable Example


```python
df = pd.DataFrame({
    "dose": ["50 mg", "75 mg", "100 mg"]
})

df["dose_mg"] = df["dose"].str.extract(r"(\d+)").astype(int)  # extract the numerical part
df
```

### Unstructured Mixed Variable Example

Approximate Numeric Values


```python
df = pd.DataFrame({
    "weight": ["~70kg", "about 65 kg", "70-ish kg", "≈75kg"]
})

df["weight_kg"] = df["weight"].str.extract(r"(\d+\.?\d*)").astype(float)
df
```

Cannot be reliably parsed using `split()`.

### Unstructured Mixed Variable Example

Duration Descriptions


```python
df = pd.DataFrame({
    "stay": ["3-5 days", "less than 2 days", "2 weeks approx", "overnight"]
})

df["value"] = df["stay"].str.extract(r"(\d+)").astype(float)
df["unit"] = df["stay"].str.extract(r"(day|week|night)", expand=False)
df
```

### Unstructured Mixed Variable Example

Clinical Notes with Measurements


```python
df = pd.DataFrame({
    "oxygen": ["SpO2 95% room air", "O2 sat: ~88% on oxygen", "92%"]
})

df["spo2"] = df["oxygen"].str.extract(r"(\d{2,3})").astype(float)
df["on_oxygen"] = df["oxygen"].str.contains("oxygen", case=False).astype(int)
df
```

### Unstructured Mixed Variable Example

Severity Embedded in Text


```python
df = pd.DataFrame({
    "pain": ["mild 2/10", "pain score ~7", "severe pain", "4-5 range"]
})

df["pain_score"] = df["pain"].str.extract(r"(\d+)").astype(float)
df["severe_flag"] = df["pain"].str.contains("severe", case=False).astype(int)
df
```
