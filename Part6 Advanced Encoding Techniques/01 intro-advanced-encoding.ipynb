# Advanced Categorical Encoding

Beyond simple encodings like one-hot and ordinal, there are **advanced encoding techniques** for categorical variables. These methods are especially useful for **high-cardinality features**, **supervised learning**, or **tree-based / neural network models**.  

We will cover:

- Target / Mean Encoding
- Leave-One-Out Encoding
- Binary / Base-N Encoding
- Hashing / Feature Hashing
- CatBoost / Monotonic Encoding

---

## 1. Target / Mean Encoding

Replace each category with the **mean (or other statistic) of the target variable** for that category.

### How It Works
Suppose `Fruit` predicts `Sales`:

| Fruit  | Sales |
|--------|-------|
| Apple  | 100   |
| Banana | 150   |
| Apple  | 120   |

**Target encoding** for `Fruit` (mean Sales):

| Fruit  | Encoded |
|--------|--------|
| Apple  | 110    |
| Banana | 150    |

### When to Use
- Supervised learning tasks
- Tree-based or linear models
- High-cardinality categorical features

### Advantages
- Captures the relationship between category and target
- Reduces dimensionality compared to one-hot

### Limitations
- Risk of **target leakage** if not carefully cross-validated
- Can overfit on small datasets

---

## 2. Leave-One-Out Encoding

Similar to target encoding but **excludes the current row** from the calculation to reduce overfitting.

### How It Works
For each row, the encoded value is the mean of the target for that category **excluding the current row**.

### When to Use
- Supervised learning
- High-cardinality features where target encoding may overfit

### Advantages
- Reduces bias compared to naive target encoding
- Still captures category-target relationship

### Limitations
- Computationally more expensive
- Needs careful cross-validation

---

## 3. Binary / Base-N Encoding

Represent each category as a **binary number** or using **log2(k)** binary columns instead of one-hot.

### How It Works
Suppose a feature has 5 categories (`A, B, C, D, E`), assign integers 0–4 and encode in binary:

| Category | Integer | Binary |
|----------|--------|--------|
| A        | 0      | 000    |
| B        | 1      | 001    |
| C        | 2      | 010    |
| D        | 3      | 011    |
| E        | 4      | 100    |

### When to Use
- High-cardinality categorical features
- Reducing dimensionality compared to one-hot

### Advantages
- Fewer columns than one-hot encoding
- Compatible with many models

### Limitations
- Introduces arbitrary numeric relationships
- May not be suitable for linear models

---

## 4. Hashing / Feature Hashing

Apply a **hash function** to categories to map them to a fixed number of numeric columns.

### How It Works
- Hash each category into a fixed-size vector
- Often used for text features or categorical features with millions of unique values

### When to Use
- Very high-cardinality features
- Memory-limited situations

### Advantages
- Fixed size regardless of category count
- Very memory-efficient

### Limitations
- Hash collisions can occur
- Not interpretable
- May require tuning number of hash bins

---

## 5. CatBoost / Monotonic Encoding

Assign **numeric values to categories** such that there is a **monotonic relationship** with the target variable. Widely used in **CatBoost and LightGBM**.

### How It Works
- Compute a target statistic (mean, sum, or other) per category
- Replace category with this value while preserving **order** or **monotonic relationship**
- Often combined with randomized permutations to prevent overfitting

### When to Use
- Tree-based boosting models (CatBoost, LightGBM)
- High-cardinality categorical features
- Regression or classification tasks

### Advantages
- Efficient for high-cardinality features
- Preserves relationship with target
- Integrated in specialized boosting libraries

### Limitations
- Specialized for boosting models
- Risk of overfitting if not regularized

---

## Summary Table

| Method | Type | Use Case | Pros | Cons |
|--------|------|---------|------|------|
| Target / Mean | Supervised | Tree/linear models, high-cardinality | Captures target relationship | Risk of leakage, overfitting |
| Leave-One-Out | Supervised | Same as above | Reduces overfitting | Computationally expensive |
| Binary / Base-N | Simple | High-cardinality | Reduces columns | Introduces arbitrary order |
| Hashing | Simple | Very high-cardinality | Memory-efficient | Collisions, not interpretable |
| CatBoost / Monotonic | Supervised | Boosting models | Preserves order with target | Specialized, may overfit |
