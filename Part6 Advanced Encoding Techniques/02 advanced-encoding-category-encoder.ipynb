```python
import pandas as pd
from sklearn.model_selection import train_test_split
import category_encoders as ce

data = {
    'Fruit': ['Apple', 'Banana', 'Apple', 'Orange', 'Apple', 'Banana', 'Orange', 'Banana', 'Apple'],
    'Size': ['Small', 'Medium', 'Large', 'Small', 'Large', 'Medium', 'Large', 'Small', 'Medium'],
    'Price': [1.2, 0.8, 1.3, 1.5, 1.1, 0.9, 1.6, 0.85, 1.25],
    'Sales': [100, 150, 120, 80, 110, 160, 70, 155, 115]
}
df = pd.DataFrame(data)
df
```

```python
X = df[['Fruit', 'Size', 'Price']]
y = df['Sales']

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)
```

## Target / Mean Encoding


```python
target_enc = ce.TargetEncoder(cols=['Fruit', 'Size'])
target_enc.fit(X_train, y_train)
X_train_encoded = target_enc.transform(X_train)
X_test_encoded = target_enc.transform(X_test)

X_train_encoded
```

## Leave-One-Out Encoding


```python
loo_enc = ce.LeaveOneOutEncoder(cols=['Fruit', 'Size'])
loo_enc.fit(X_train, y_train)
X_train_loo = loo_enc.transform(X_train)
X_test_loo = loo_enc.transform(X_test)

X_train_loo
```

## Binary / Base-N Encoding


```python
binary_enc = ce.BinaryEncoder(cols=['Fruit', 'Size'])
binary_enc.fit(X_train)
X_train_bin = binary_enc.transform(X_train)
X_test_bin = binary_enc.transform(X_test)

X_train_bin
```

## Hashing / Feature Hashing


```python
hash_enc = ce.HashingEncoder(cols=['Fruit', 'Size'], n_components=3)
hash_enc.fit(X_train)
X_train_hash = hash_enc.transform(X_train)
X_test_hash = hash_enc.transform(X_test)

X_train_hash
```

## CatBoost / Monotonic Encoding


```python
catboost_enc = ce.CatBoostEncoder(cols=['Fruit', 'Size'])
catboost_enc.fit(X_train, y_train)
X_train_cb = catboost_enc.transform(X_train)
X_test_cb = catboost_enc.transform(X_test)

X_train_cb
```
