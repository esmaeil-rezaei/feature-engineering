# Discretization

**Discretization** is the process of converting continuous numerical variables into discrete categories or intervals (bins). It is commonly used in preprocessing to improve model interpretability, reduce noise, and meet the assumptions of certain algorithms such as decision trees and rule-based models.

---

## Equal-Width Discretization

**Equal-width discretization** divides the range of a variable into intervals of equal size.

- **Method:**  
  If a variable ranges from `min` to `max` and is divided into `k` bins, each bin has width
  $$
  \text{Width} = \frac{\max - \min}{k}
  $$
- **Pros:** Simple and fast to compute.
- **Cons:** Sensitive to outliers and may produce empty or unbalanced bins.

---

## Equal-Frequency Discretization

**Equal-frequency discretization** (also called quantile-based discretization) creates bins such that each bin contains approximately the same number of observations.

- **Method:** Sort the data and divide it into `k` bins with equal counts.
- **Pros:** Ensures balanced bins.
- **Cons:** Bin widths may vary significantly and can split identical values across bins.

---

## Arbitrary Discretization

**Arbitrary discretization** uses manually defined cut points based on domain knowledge or business rules.

- **Example:** Age groups such as `0–18`, `19–35`, `36–60`, `60+`.
- **Pros:** Highly interpretable and domain-driven.
- **Cons:** Subjective and may not generalize well across datasets.

---

## Discretization Using K-Means

**K-means discretization** applies clustering to group continuous values into `k` clusters.

- **Method:**
  - Apply k-means clustering to a single continuous feature.
  - Use cluster assignments as discrete categories.
- **Pros:** Data-driven and adapts to the distribution.
- **Cons:** Requires specifying `k` and is sensitive to initialization.

---

## Discretization Using Decision Trees

**Decision tree-based discretization** uses a supervised learning approach to find optimal split points.

- **Method:**
  - Train a decision tree using the continuous feature.
  - Extract split thresholds as discretization boundaries.
- **Pros:** Optimized for predictive performance and target-aware.
- **Cons:** Requires labeled data and may overfit without regularization.

---

## Binarization

**Binarization** transforms a continuous feature into a binary variable based on a threshold.

- **Method:**
  $$
  x' =
  \begin{cases}
  1, & x \geq \text{threshold} \\
  0, & x < \text{threshold}
  \end{cases}
  $$
- **Pros:** Very simple and interpretable.
- **Cons:** Loses information by reducing values to two categories.
