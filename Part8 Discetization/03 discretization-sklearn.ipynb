```python
import numpy as np
from sklearn.preprocessing import KBinsDiscretizer

X = np.array([[10], [15], [20], [25], [30], [35], [40], [45]])
```

### Equal-Width Discretization


```python
ew = KBinsDiscretizer(
    n_bins=3,
    encode="ordinal",
    strategy="uniform"   # equal-width
)

X_ew = ew.fit_transform(X)
X_ew
```

### Equal-Frequency Discretization


```python
ef = KBinsDiscretizer(
    n_bins=3,
    encode="ordinal",
    strategy="quantile"  # equal-frequency
)

X_ef = ef.fit_transform(X)
X_ef
```

### Arbitrary Discretization (Manual Bins)


```python
bins = [0, 20, 35, 50]  # domain-driven cut points
X_arbitrary = np.digitize(X.flatten(), bins)

X_arbitrary
```

### Discretization Using K-Means


```python
km = KBinsDiscretizer(
    n_bins=3,
    encode="ordinal",
    strategy="kmeans"
)

X_km = km.fit_transform(X)
X_km
```

### Discretization Using Decision Tree


```python
from sklearn.tree import DecisionTreeClassifier

y = np.array([0, 0, 0, 1, 1, 1, 1, 1])

dt = DecisionTreeClassifier(max_depth=2)
dt.fit(X, y)

thresholds = dt.tree_.threshold
thresholds = thresholds[thresholds != -2]  # valid splits

print("Split thresholds:", thresholds)
```

### Binarization


```python
from sklearn.preprocessing import Binarizer

binz = Binarizer(threshold=30)
X_bin = binz.fit_transform(X)

X_bin
```
