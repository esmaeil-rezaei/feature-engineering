```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "x": [10, 15, 20, 25, 30, 35, 40, 45]
})

df
```

### Equal-Width Discretization


```python
df["x_eq_width"] = pd.cut(df["x"], bins=3, labels=False, include_lowest=True)
df
```

Remove `labels=False` to show the intervals.


```python
df["x_eq_width_intervals"] = pd.cut(df["x"], bins=3, include_lowest=True)
df
```

### Equal-Frequency Discretization


```python
df["x_eq_freq"] = pd.qcut(df["x"], q=3, labels=False)
df
```

### Arbitrary Discretization


```python
bins = [0, 20, 35, 50]
labels = ["low", "medium", "high"]

df["x_arbitrary"] = pd.cut(df["x"], bins=bins, labels=labels)
df
```

### Binarization


```python
df["x_binary"] = (df["x"] >= 30).astype(int)
df
```
