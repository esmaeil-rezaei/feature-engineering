```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer

df = pd.DataFrame({
    'age': [25, 32, None, 40, 28],
    'income': [50000, 60000, 55000, None, 70000],
    'gender': ['female', None, None, 'male', 'male']
})

df
```

```python
# Median imputer
median_imputer = SimpleImputer(strategy='median')
median_imputer.fit_transform(df[['age']])
```

```python
# Mean imputer
median_imputer = SimpleImputer(strategy='mean')
median_imputer.fit_transform(df[['age']])
```

```python
# Fill missing values with -999
arbitrary_imputer = SimpleImputer(strategy='constant', fill_value=-999)
arbitrary_imputer.fit_transform(df[['income']])
```

```python
# Fill missing categorical values with 'female'
cat_arbitrary_imputer = SimpleImputer(strategy='constant', fill_value='female')
cat_arbitrary_imputer.fit_transform(df[['gender']])
```

`SimpleImputer` does not recognize `None` as a missing value for object arrays unless you explicitly tell it using the `missing_values` parameter.


```python
cat_arbitrary_imputer = SimpleImputer(strategy='constant', fill_value='female', missing_values=None)
cat_arbitrary_imputer.fit_transform(df[['gender']])
```

```python
# Mode / Most Frequent Imputation
mode_imputer = SimpleImputer(strategy='most_frequent')
mode_imputer.fit_transform(df[['age']])
```

```python
# Complete Case Analysis (CCA)
from sklearn.preprocessing import FunctionTransformer

drop_rows = FunctionTransformer(lambda X: X.dropna(), validate=False)
df_clean = drop_rows.transform(df)
df_clean
```

```python
# End of Distribution Imputation
# Replace missing numeric values with max + 1000
end_of_dist_imputer = SimpleImputer(strategy='constant', fill_value=df['income'].max() + 1000)
end_of_dist_imputer.fit_transform(df[['income']])
```

```python
# Random Sample Imputation
from sklearn.preprocessing import FunctionTransformer

def random_sample_impute(X):
    X = X.copy()
    for col in X.columns:
        na_mask = X[col].isna()
        X.loc[na_mask, col] = np.random.choice(X.loc[~na_mask, col], size=na_mask.sum())
    return X

random_imputer = FunctionTransformer(random_sample_impute, validate=False)
random_imputer.transform(df)
```

```python
from sklearn.impute import MissingIndicator

indicator = MissingIndicator()
indicator.fit_transform(df[['age', 'income']])
```
