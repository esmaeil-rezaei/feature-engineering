# Categorical Encoding in Pandas

Pandas provides simple and efficient tools to encode categorical variables directly, without relying on external libraries. Below, we demonstrate **One-Hot Encoding** and **Ordinal Encoding** using pure pandas workflows.

## One-Hot Encoding in Pandas


```python
import pandas as pd

df = pd.DataFrame({
    'Color': ['Red', 'Blue', 'Green', 'Red']
})

df
```

```python
df_ohe = pd.get_dummies(df, columns=['Color'])
df_ohe
```

```python
# Dropping One Dummy Variable (Avoid Multicollinearity)
df_ohe_drop = pd.get_dummies(df, columns=['Color'], drop_first=True)
df_ohe_drop
```

## Ordinal Encoding in Pandas


```python
df = pd.DataFrame({
    'Education': ['High School', 'Bachelor', 'Master', 'PhD']
})

df
```

```python
# Ordinal Encoding Using a Mapping Dictionary
education_order = {
    'High School': 0,
    'Bachelor': 1,
    'Master': 2,
    'PhD': 3
}

df['Education_encoded'] = df['Education'].map(education_order)
df
```

## Count Encoding


```python
data = {
    'Fruit': ['Apple', 'Banana', 'Apple', 'Orange', 'Apple', 'Banana', 'Orange', 'Banana', 'Apple']
}
df = pd.DataFrame(data)
df
```

```python
# Count encoding
df['Fruit_Count'] = df['Fruit'].map(df['Fruit'].value_counts())
print("\nCount Encoded Data:")
df
```

```python
# Frequency encoding
df['Fruit_Freq'] = df['Fruit'].map(df['Fruit'].value_counts(normalize=True))
print("\nFrequency Encoded Data:")
df
```
