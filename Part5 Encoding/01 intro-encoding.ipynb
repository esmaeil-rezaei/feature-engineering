# Categorical Encoding

Many real-world datasets contain **categorical variables**, which represent qualitative information such as colors, education levels, or product types. Since most machine learning algorithms operate on numerical data, categorical variables must be transformed into a numerical format. This process is known as **categorical encoding**.

Two of the most commonly used encoding techniques are **One-Hot Encoding** and **Ordinal Encoding**. Each method is designed for different types of categorical variables and carries different assumptions about the data.

---

## One-Hot Encoding

**One-Hot Encoding** transforms a categorical variable with _k_ unique categories into _k_ binary (0/1) features. Each feature represents the presence or absence of a specific category.

### How It Works

Suppose we have a categorical variable `Color`:

| Color |
| ----- |
| Red   |
| Blue  |
| Green |

After one-hot encoding:

| Color_Red | Color_Blue | Color_Green |
| --------- | ---------- | ----------- |
| 1         | 0          | 0           |
| 0         | 1          | 0           |
| 0         | 0          | 1           |

Each row contains exactly one `1`, indicating the observed category.

### When to Use

- Nominal (unordered) categories
- Categories with no intrinsic ranking
- Algorithms that assume linear relationships (e.g., linear regression, logistic regression)

### Advantages

- No artificial ordering is introduced
- Works well with most machine learning models
- Easy to interpret

### Limitations

- Increases dimensionality, especially with high-cardinality features
- Can lead to sparse feature matrices
- May cause multicollinearity if all dummy variables are retained

---

## Label Encoding

**Label Encoding** converts each category into a unique integer. Unlike One-Hot Encoding, it produces a single column of numeric values.

### How It Works

Suppose we have a categorical target variable `Animal`:

| Animal |
| ------ |
| Cat    |
| Dog    |
| Fish   |
| Cat    |

After label encoding:

| Animal | Encoded Value |
| ------ | ------------- |
| Cat    | 0             |
| Dog    | 1             |
| Fish   | 2             |
| Cat    | 0             |

- Internally, the labels are **sorted alphabetically** before assignment.
- This encoding is **mostly used for target labels (y)**, not features, because the numbers are arbitrary and do not imply order.

### When to Use

- Target variables in classification problems
- When models require numeric labels
- Quick numeric representation for simple algorithms
- Order does not matter

### Advantages

- Simple and compact
- Preserves mapping to original labels via inverse transformation
- Easy to implement

### Limitations

- Numbers may imply order if misused on features
- Not suitable for nominal feature columns in linear models
- **Mapping is not guaranteed to be consistent across runs or datasets** (categories may be assigned different numbers)

---

## Ordinal Encoding

**Ordinal Encoding** assigns an integer value to each category based on a predefined order. The numerical values reflect the ranking or hierarchy among categories.

### How It Works

Suppose we have an ordered variable `Education Level`:

| Education Level |
| --------------- |
| High School     |
| Bachelor        |
| Master          |
| PhD             |

After ordinal encoding:

| Education Level | Encoded Value |
| --------------- | ------------- |
| High School     | 0             |
| Bachelor        | 1             |
| Master          | 2             |
| PhD             | 3             |

The encoded values preserve the natural order of the categories.

### When to Use

- Ordinal variables with a clear and meaningful ranking
- Tree-based models (e.g., decision trees, random forests, gradient boosting)
- Situations where relative ordering carries semantic meaning

### Advantages

- Compact representation (single feature)
- Preserves category order
- Computationally efficient

### Limitations

- Imposes a numeric distance that may not reflect true differences
- Can mislead linear models if the spacing between categories is not uniform
- Requires correct domain knowledge to define the order

---

## Count / Frequency Encoding

**Count Encoding** (also called **Frequency Encoding**) replaces each category with the number of times it appears (count) or its proportion (frequency) in the dataset.

### How It Works

Suppose we have a categorical variable `Fruit`:

| Fruit  |
| ------ |
| Apple  |
| Banana |
| Apple  |
| Orange |
| Apple  |
| Banana |

**Count encoding:**

| Fruit  | Count |
| ------ | ----- |
| Apple  | 3     |
| Banana | 2     |
| Apple  | 3     |
| Orange | 1     |
| Apple  | 3     |
| Banana | 2     |

**Frequency encoding:**

| Fruit  | Frequency |
| ------ | --------- |
| Apple  | 0.5       |
| Banana | 0.33      |
| Apple  | 0.5       |
| Orange | 0.17      |
| Apple  | 0.5       |
| Banana | 0.33      |

### When to Use

- High-cardinality categorical features (many unique categories)
- When one-hot encoding would lead to very large sparse matrices
- For tree-based or linear models that can handle numeric inputs

### Advantages

- Reduces dimensionality compared to one-hot encoding
- Captures information about category prevalence
- Easy to implement

### Limitations

- Does not preserve order between categories
- May introduce bias if frequencies correlate with the target variable
- Can be misleading for small datasets with rare categories

---

_To avoid multicolliniarity, it is common to drop one of the columns for categories. For exaxmple, we drop `Blue` color._

You might wonder whether dropping a dummy variable causes a loss of information for the Blue category. However, no information is lost when one dummy column is removed, provided the model includes an intercept (bias) term. When `Red = 0` and `Blue = 0`, the observation belongs to the Green category. In this formulation, the intercept represents the baseline effect of Green, while the coefficients of Red and Blue represent differences relative to Green.

#### ML Models Where You SHOULD Drop One Dummy Column

These models assume linearity or additivity and are sensitive to multicollinearity.

**Linear Models**

- Linear Regression

- Ridge Regression

- Lasso Regression

- Elastic Net

**Generalized Linear Models (GLMs)**

- Logistic Regression

- Multinomial Logistic Regression

- Poisson Regression

- Cox Proportional Hazards Model

**Statistical / Interpretable Models**

- Probit Models

- GAMs (Generalized Additive Models)

- Fixed-effects / Mixed-effects models
